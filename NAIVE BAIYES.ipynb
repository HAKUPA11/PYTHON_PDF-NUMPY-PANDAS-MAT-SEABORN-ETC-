```python
from sklearn.datasets import fetch_openml
breast_cancer=fetch_openml(name="breast-cancer", version=1, as_frame=True)
df=breast_cancer.frame
print("ORIGINAL DATA")

print(df.head(), "\n")
print(df.tail(), "\n")
print(df.info(), "\n")
print(df.describe(), "end=\n")
```

```python
#USING LABEL ENCODER
le=LabelEncoder()

df["age"] = le.fit_transform(df["age"])
df["menopause"] = le.fit_transform(df["menopause"])
df["node-caps"] = le.fit_transform(df["node-caps"])

X = df[["age", "menopause"]]
y=df["node-caps"]
```

```python
#TRAIN TEST SPLIT
X_train, X_test, y_train, y_test=train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
model=GaussianNB()
model.fit(X_test, y_test)
y_predict=model.predict(X_test)
```

```python
conm=confusion_matrix(y_test, y_predict )
print("CONFUSION MATRIX : \n", conm)
print("\n")

acc_score=accuracy_score(y_test, y_predict)
print("ACCURACY SCORE :", acc_score)

print("\nClassification Report:")
print(classification_report(y_test, y_predict))
```
