we select the centroid hich are the centre points of different groups and then we draw a line joing these centroids and after wards a perpendicular line on this joint line



now every point above the perpendicular line and near the red point(centroid) will be turned red and every points below near the yellow centroid will be turned yellow



now we calculate the mean of all the data points (different colours) via this we will get a new centroid    and the mean be towards that area where the density of the points are more and therefore the new centroid will also be towards that 
and then we repeat the second point and we again mark the red points and the yellow points then we again repeat the third point we will keep on doing this until there is no movement of centroid because by that we would have divided the groups accurately
